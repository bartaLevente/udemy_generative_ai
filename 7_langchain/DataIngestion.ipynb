```python
from langchain_community.document_loaders import TextLoader

textloader = TextLoader('speech.txt')
text_documents = textloader.load()

print(text_documents)
```

```python
from langchain_community.document_loaders import PyPDFLoader

pdf_loader = PyPDFLoader('attention.pdf')
pdf_documents = pdf_loader.load()

print(pdf_documents)
print(len(pdf_documents))
```

```python
from langchain_community.document_loaders import WebBaseLoader
import bs4

webloader = WebBaseLoader(web_paths=["https://medium.com/@sahin.samia/what-is-reranking-in-retrieval-augmented-generation-rag-ee3dd93540ee"],
bs_kwargs=dict(parse_only=bs4.SoupStrainer(["h1", "h2", "h3"])))

web_documents = webloader.load()
print(web_documents)
```

```python
from langchain_community.document_loaders import ArxivLoader
arxiv_docs = ArxivLoader(query="2305.14024",load_max_docs=2).load()
print(arxiv_docs)
print(len(arxiv_docs))
```

```python
from langchain_community.document_loaders import WikipediaLoader

wikipedia_loader = WikipediaLoader(query="Artificial Intelligence", load_max_docs=2)
wikipedia_documents = wikipedia_loader.load()

print(wikipedia_documents)
print(len(wikipedia_documents))
```
