```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"] = os.getenv("OPENAI_API_KEY")
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
from langchain_text_splitters import CharacterTextSplitter

embeddings = OpenAIEmbeddings(model="text-embedding-3-small")

textloader = TextLoader('speech.txt')
text_documents = textloader.load()

text_splitter = CharacterTextSplitter(separator="\n\n", chunk_size=100, chunk_overlap=20)
split_text_documents = text_splitter.split_documents(text_documents)

vectorstore = Chroma.from_documents(split_text_documents, embeddings)
```

```python
query = "We have borne with their present government through all these bitter months because of that friendship—exercising"
results = vectorstore.similarity_search(query, k=2)
print("Results:")
for result in results:
    print(result.page_content, result.metadata)
```

```python
##Ollama embeddings
from langchain_community.embeddings import OllamaEmbeddings

ollama_embeddings = OllamaEmbeddings(model="gemma:2b")
ollama_embedded_texts = ollama_embeddings.embed_documents(split_text_documents)
print("Ollama Embedded Texts:")
for embedded_text in ollama_embedded_texts[:3]:
    print(embedded_text)
```
