# Faiss

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import FAISS
from langchain_community.embeddings import OllamaEmbeddings
from langchain_text_splitters import CharacterTextSplitter

textloader = TextLoader('speech.txt')
text_documents = textloader.load()
text_splitter = CharacterTextSplitter(separator="\n\n", chunk_size=1000, chunk_overlap=30)
docs = text_splitter.split_documents(text_documents)


embeddings = OllamaEmbeddings(model="gemma:2b")
db=FAISS.from_documents(docs, embeddings)
db
```

```python
query = "How does the speaker describe the outcom eof the war?"
result = db.similarity_search(query, k=3)
for doc in result:
    print(doc.page_content)
    print("===")
```

```python
### Using the retriever interface
retriever = db.as_retriever()
retrieval_result = retriever.invoke(query)
for doc in retrieval_result:
    print(doc.page_content)
    print("===")
```

```python
with_scores = db.similarity_search_with_score(query, k=3)
for doc, score in with_scores:
    print(f"Score: {score}")
    print(doc.page_content)
    print("===")
```

```python
db.save_local("faiss_index")

new_df = FAISS.load_local("faiss_index", embeddings,allow_dangerous_deserialization=True)
```
